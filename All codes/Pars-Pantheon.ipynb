```python
def sigma_dl(dL):
    
    sigma_dl = dL * 0.22 * np.log(10)/5
    return sigma_dl

def chi_square(theta,func, z, data, order=order):
    sigma = sigma_dl(data)
    diff2 = (func(z, *theta, order=order) - data)**2

    chi = np.sum(diff2/sigma**2)
    return chi
```
