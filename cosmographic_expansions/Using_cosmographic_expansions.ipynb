```python
H0_range: Tuple[float, float] = (50.0, 90.0)
q0_range: Tuple[float, float] = (-10.0, 10.0)
j0_range: Tuple[float, float] = (-10.0, 10.0)
s0_range: Tuple[float, float] = (-10.0, 10.0)
c0_range: Tuple[float, float] = (-10.0, 20.0)
Ok_range: Tuple[float, float] = (0.0, 0.0)

# Convert cosmographic parameter ranges to EIS parameter ranges
# Using the middle values of the ranges to get representative EIS ranges
H0_mid = (H0_range[0] + H0_range[1]) / 2
q0_mid = (q0_range[0] + q0_range[1]) / 2
j0_mid = (j0_range[0] + j0_range[1]) / 2
s0_mid = (s0_range[0] + s0_range[1]) / 2
c0_mid = (c0_range[0] + c0_range[1]) / 2
p0_mid = (p0_range[0] + p0_range[1]) / 2
Ok_mid = (Ok_range[0] + Ok_range[1]) / 2

# Get EIS values at the extremes to determine ranges
E1_min, E2_min, E3_min, E4_min, E5_min = EISTransforms.cosmographic_to_eis(q0_range[0], j0_range[0], s0_range[0], c0_range[0], p0_range[0], Ok_range[0])
E1_max, E2_max, E3_max, E4_max, E5_max = EISTransforms.cosmographic_to_eis(q0_range[1], j0_range[1], s0_range[1], c0_range[1], p0_range[1], Ok_range[1])

E1_range: Tuple[float, float] = (min(E1_min, E1_max), max(E1_min, E1_max))
E2_range: Tuple[float, float] = (min(E2_min, E2_max), max(E2_min, E2_max))
E3_range: Tuple[float, float] = (min(E3_min, E3_max), max(E3_min, E3_max))
E4_range: Tuple[float, float] = (min(E4_min, E4_max), max(E4_min, E4_max))
E5_range: Tuple[float, float] = (min(E5_min, E5_max), max(E5_min, E5_max))
```

```python
from astropy.cosmology import FlatLambdaCDM

dL = ExpansionModels.z_series(z, H0, q0, j0, s0, c0, p0, Ok, order=3)

import astropy.units as u

# Convert cosmographic parameters to approximate cosmological parameters
# For a flat LCDM model with q0 = -0.55, this corresponds to Omega_m ~ 0.3
cosmo = FlatLambdaCDM(H0=H0, Om0=0.3)
dL_astropy = cosmo.luminosity_distance(z).value  # in Mpc

```

```python
zs = np.linspace(0.01,5,100)
plt.figure(figsize= (10, 6))
for order in range(1, 7):
    plt.plot(zs, ExpansionModels.z_series(zs, H0, q0, j0, s0, c0, p0, Ok, order=order), label=f'z order={order}')
plt.plot(zs, cosmo.luminosity_distance(zs).to(u.Mpc).value, label='Astropy',color='k', linestyle='--')
plt.xlabel('Redshift z')
plt.ylabel('Luminosity Distance (Mpc)')
plt.legend()
plt.grid(True)
plt.title('Cosmographic Luminosity Distance Approximations (z series)')

plt.figure(figsize=(10, 6))

plt.title('Cosmographic Luminosity Distance Approximations (log series)')
for order in range(1, 7):
    plt.plot(zs, ExpansionModels.log_series(np.log(1 + zs), H0, q0, j0, s0, c0, p0, Ok, order=order), label=f'log order={order}')
plt.plot(zs, cosmo.luminosity_distance(zs).to(u.Mpc).value, label='Astropy', color='k', linestyle='--')
plt.xlabel('Redshift z')
plt.ylabel('Luminosity Distance (Mpc)')
plt.legend()
plt.grid(True)
plt.figure(figsize=(10, 6))
plt.title('Cosmographic Luminosity Distance Approximations (y series)')
for order in range(1, 7):
    plt.plot(zs, ExpansionModels.y_series(zs, H0, q0, j0, s0, c0, p0, Ok, order=order), label=f'y order={order}')
plt.plot(zs, cosmo.luminosity_distance(zs).to(u.Mpc).value, label='Astropy', color='k',linestyle='--')
plt.xlabel('Redshift z')
plt.ylabel('Luminosity Distance (Mpc)')
plt.legend()
plt.grid(True)
plt.figure(figsize= (10, 6))

plt.title('Cosmographic Luminosity Distance Approximations(Pade)')
plt.plot(zs, PadeApproximants.pade_21(zs, H0, q0, j0, s0, c0, p0, Ok, order=2), label='Pade 2/1',alpha=0.8)
plt.plot(zs, PadeApproximants.pade_22(zs, H0, q0, j0, s0, c0, p0, Ok, order=2), label='Pade 2/2',alpha=0.8)
plt.plot(zs, PadeApproximants.pade_32(zs, H0, q0, j0, s0, c0, p0, Ok, order=3), label='Pade 3/2',alpha=0.8)
plt.plot(zs, PadeApproximants.pade_42(zs, H0, q0, j0, s0, c0, p0, Ok, order=4), label='Pade 4/2',alpha=0.8)
plt.plot(zs, PadeApproximants.pade_average(zs, H0, q0, j0, s0, c0, p0, Ok, order=3), label='Pade Average',alpha=0.8)
plt.plot(zs,PadeApproximants.pade_51(zs,H0,q0,j0,s0,c0,p0,Ok,order=5), label='Pade 5/1',alpha=0.8)
plt.plot(zs, cosmo.luminosity_distance(zs).to(u.Mpc).value, label='Astropy',color='k', linestyle='--')
plt.xlabel('Redshift z')
plt.ylabel('Luminosity Distance (Mpc)')
plt.legend()
plt.grid(True)

plt.figure(figsize=(10, 6))
plt.title('EIS Luminosity Distance Approximations')
for order in range(1, 7):
    
    dL_eis = EISModel.luminosity_distance(zs, H0, E1, E2, E3, E4, E5, order=order)
    plt.plot(zs, dL_eis, label=f'EIS order={order}')
plt.plot(zs, cosmo.luminosity_distance(zs).to(u.Mpc).value, label='Astropy', color='k', linestyle='--')
plt.xlabel('Redshift z')
plt.ylabel('Luminosity Distance (Mpc)')
plt.legend()
plt.grid(True)
```

```python
def relative_error(model, reference):
    return 100 * (model - reference) / reference

ref = cosmo.luminosity_distance(zs).to(u.Mpc).value

plt.figure(figsize=(10, 6))
for order in range(1, 6):
    model = ExpansionModels.z_series(zs, H0, q0, j0, s0, c0, p0, Ok, order=order)
    plt.plot(zs, relative_error(model, ref), label=f'z order={order}')
plt.axhline(0, color='k', linestyle='--', linewidth=1)
plt.xlabel('Redshift z')
plt.ylabel('Relative Error (%)')
plt.legend()
plt.grid(True)
plt.title('Relative Error: Cosmographic z Series vs Astropy')

plt.figure(figsize=(10, 6))
for order in range(1, 6):
    model = ExpansionModels.log_series(np.log(1 + zs), H0, q0, j0, s0, c0, p0, Ok, order=order)
    plt.plot(zs, relative_error(model, ref), label=f'log order={order}')
plt.axhline(0, color='k', linestyle='--', linewidth=1)
plt.xlabel('Redshift z')
plt.ylabel('Relative Error (%)')
plt.legend()
plt.grid(True)
plt.title('Relative Error: Cosmographic log Series vs Astropy')

plt.figure(figsize=(10, 6))
for order in range(1, 6):
    model = ExpansionModels.y_series(zs, H0, q0, j0, s0, c0, p0, Ok, order=order)
    plt.plot(zs, relative_error(model, ref), label=f'y order={order}')
plt.axhline(0, color='k', linestyle='--', linewidth=1)
plt.xlabel('Redshift z')
plt.ylabel('Relative Error (%)')
plt.legend()
plt.grid(True)
plt.title('Relative Error: Cosmographic y Series vs Astropy')

plt.figure(figsize=(10, 6))
plt.plot(zs, relative_error(PadeApproximants.pade_21(zs, H0, q0, j0, s0, c0, p0, Ok, order=2), ref), label='Pade 2/1', alpha=0.8)
plt.plot(zs, relative_error(PadeApproximants.pade_22(zs, H0, q0, j0, s0, c0, p0, Ok, order=2), ref), label='Pade 2/2', alpha=0.8)
plt.plot(zs, relative_error(PadeApproximants.pade_32(zs, H0, q0, j0, s0, c0, p0, Ok, order=3), ref), label='Pade 3/2', alpha=0.8)
plt.plot(zs, relative_error(PadeApproximants.pade_42(zs, H0, q0, j0, s0, c0, p0, Ok, order=4), ref), label='Pade 4/2', alpha=0.8)
plt.plot(zs, relative_error(PadeApproximants.pade_average(zs, H0, q0, j0, s0, c0, p0, Ok, order=3), ref), label='Pade Average', alpha=0.8)
plt.plot(zs, relative_error(PadeApproximants.pade_51(zs, H0, q0, j0, s0, c0, p0, Ok, order=5), ref), label='Pade 5/1', alpha=0.8)
plt.axhline(0, color='k', linestyle='--', linewidth=1)
plt.xlabel('Redshift z')
plt.ylabel('Relative Error (%)')
plt.legend()
plt.grid(True)
plt.title('Relative Error: Pade Approximants vs Astropy')

plt.figure(figsize=(10, 6))
for order in range(1, 6):
    model = EISModel.luminosity_distance(zs, H0, E1, E2, E3, E4, E5, order=order)
    plt.plot(zs, relative_error(model, ref), label=f'EIS order={order}')
plt.axhline(0, color='k', linestyle='--', linewidth=1)
plt.xlabel('Redshift z')
plt.ylabel('Relative Error (%)')
plt.legend()
plt.grid(True)
plt.title('Relative Error: EIS Model vs Astropy')
```
